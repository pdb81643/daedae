```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
print(iris)
```

```python
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
```

```python
print(df)
```

```python
df['species']=iris.target
```

```python
plt.plot(df['sepal length (cm)'])
```

```python
df['species'] = df['species'].map({0: "품종1", 1: "품종2", 2: "품종3"})
```

```python
df.plot(kind="scatter", x='sepal length (cm)', y='petal length (cm)')
plt.grid()
```

```python
import seaborn as sns
```

```python
g = sns.FacetGrid(df, hue='species', height=6)
g.map(plt.scatter, 'sepal length (cm)', 'petal length (cm)')
g.add_legend()
```

```python
fig, axes = plt.subplots(1, 2, sharey=True, figsize=(10, 4))
```

```python
_, axes = plt.subplots(1, 2, sharey=True, figsize=(10, 4))
sns.boxplot(x='petal length (cm)', y='species', data=df, ax=axes[0])
sns.violinplot(x='petal length (cm)', y='species', data=df, ax=axes[1])
```
